# Data Processing and cleaning

```python
#combine relavant features into a single text column
def clean_text(row):
  title = row['title'] if pd.notna(row['title']) else ''
  brand = row['brand'] if pd.notna(row['brand']) else ''
  category = row['category'] if pd.notna(row['category']) else ''
  description = row['description'] if pd.notna(row['description']) else ''
  specs = ' '.join([f"{k}:{v}" for k, v in eval(row['specifications']).items()]) if pd.notna(row['specifications']) else ""
  return f"{title} {brand} {category} {specs}"

df['clean_text'] = df.apply(clean_text, axis=1)
df = df.dropna(subset=['clean_text']) #drop rows without description
```

để gộp các thuộc tính quan trọng của sản phẩm (tiêu đề, thương hiệu, danh mục, thông số kỹ thuật,...) từ một dòng dữ liệu (row) trong DataFrame thành một chuỗi văn bản duy nhất (clean_text). Chuỗi này thường được dùng làm đầu vào cho các mô hình NLP, tính toán độ tương đồng (cosine similarity), hoặc tạo vector embedding cho hệ thống tìm kiếm/gợi ý sản phẩm.

title = row['title'] if pd.notna(row['title']) else ''
Ý nghĩa: Lấy giá trị của cột title (tiêu đề sản phẩm). Hàm pd.notna() kiểm tra xem giá trị có bị trống (NaN / None) hay không. Nếu có dữ liệu thì lấy, ngược lại trả về chuỗi rỗng '' để tránh lỗi khi ghép chuỗi.

Ý nghĩa: Xử lý cột specifications (thông số kỹ thuật, thường được lưu dưới dạng chuỗi chứa dictionary Python, ví dụ: "{'Color': 'Red', 'Size': 'M'}").

pd.notna(...): Kiểm tra cột specifications có dữ liệu không.

eval(...): Chuyển đổi chuỗi văn bản thành một dictionary thực sự trong Python.

.items(): Lấy danh sách các cặp khóa-giá trị (key, value).

List comprehension [f"{k}:{v}" for k, v in ...] chuyển mỗi cặp thành chuỗi dạng "Key:Value".

' '.join(...): Nối tất cả các thông số kỹ thuật thành một chuỗi duy nhất, cách nhau bằng khoảng trắng. Nếu không có dữ liệu, trả về chuỗi rỗng "".

return f"{title} {brand} {category} {specs}"
Ý nghĩa: Ghép các thuộc tính title, brand, category và specs lại với nhau thành một chuỗi văn bản hoàn chỉnh, cách nhau bởi khoảng trắng, rồi trả về kết quả cho hàm.

df['clean_text'] = df.apply(clean_text, axis=1)
Ý nghĩa:

df.apply(..., axis=1): Áp dụng hàm clean_text lên từng hàng (axis=1) của DataFrame df.

Tạo một cột mới tên là clean_text chứa chuỗi văn bản tổng hợp của từng sản phẩm.

df = df.dropna(subset=['clean_text']) #drop rows without description
Ý nghĩa: Loại bỏ (dropna) các dòng mà cột clean_text bị trống (NaN), đảm bảo tập dữ liệu sạch sẽ trước khi đưa vào các bước huấn luyện mô hình phía sau.

# Create Embeedings

sentence-transformers là thư viện cực kỳ phổ biến và mạnh mẽ trong lĩnh vực Xử lý ngôn ngữ tự nhiên (NLP), chuyên dùng để:

Chuyển đổi văn bản thành Vector (Embedding): Biến các câu, đoạn văn, hoặc cột clean_text của bạn thành các vector số học (dense vectors) giữ nguyên ý nghĩa ngữ nghĩa.

Xây dựng hệ thống tìm kiếm ngữ nghĩa (Semantic Search): So sánh độ tương đồng giữa câu truy vấn của người dùng và danh sách sản phẩm dựa trên ý nghĩa thay vì chỉ khớp từ khóa chính xác.

Gợi ý sản phẩm (Recommendation Systems): Tính toán độ tương đồng Cosine Similarity giữa các vector sản phẩm để tìm ra các sản phẩm giống nhau.

```python
# Cài đặt thư viện sentence-transformers
# !pip install -q sentence-transformers

from sentence_transformers import SentenceTransformer

# load a pre-trained sentence-bert model
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')

# generate embeddings for product descriptions
df['embeddings'] = df['clean_text'].apply(lambda x: model.encode(x))
```

tải mô hình AI (Sentence-BERT) và chuyển đổi các chuỗi văn bản (clean_text) thành các vector biểu diễn (embeddings) để máy tính hiểu được ý nghĩa ngữ nghĩa, sau đó lưu kết quả vào cột mới df['embeddings'].

Import thư viện và tải mô hình all-MiniLM-L6-v2 từ Hugging Face. Đây là một mô hình Sentence-BERT rất phổ biến, nhẹ, chạy nhanh và tạo ra các vector có chiều dài 384 chiều, giữ lại ngữ nghĩa rất tốt cho các bài toán tìm kiếm/gợi ý.

df['embeddings'] = df['clean_text'].apply(lambda x: model.encode(x))
Ý nghĩa: Duyệt qua từng dòng trong cột clean_text, dùng hàm model.encode(x) để biến câu văn bản thành một mảng vector số thực, rồi lưu vào cột df['embeddings']

# Recommendation Engine

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

def recommend_products(query, top=5):
  query = model.encode(query)
  # Đảm bảo reshape query và vector x cho đúng định dạng 2D
  df['similarity'] = df['embeddings'].apply(lambda x: cosine_similarity([query], [x])[0][0])
  recommendations = df.sort_values(by='similarity', ascending=False).head(top)
  return recommendations[['title', 'brand', 'category', 'similarity']]
```

để xây dựng một hàm tìm kiếm và gợi ý sản phẩm dựa trên ngữ nghĩa (Semantic Search & Recommendation) sử dụng độ tương đồng Cosine (cosine_similarity). Khi người dùng gõ một từ khóa hoặc câu truy vấn bất kỳ, hàm sẽ tìm ra những sản phẩm có nội dung phù hợp nhất.

Ý nghĩa: Định nghĩa hàm recommend_products nhận vào 2 tham số: query (câu truy vấn/từ khóa người dùng tìm kiếm, ví dụ: "smartphone pin trâu" hoặc "giày chạy bộ nam") và top (số lượng sản phẩm gợi ý muốn trả về, mặc định là 5).

Dùng model.encode(query) để chuyển câu truy vấn văn bản của người dùng thành một vector số học (có cùng không gian chiều với các vector sản phẩm trong cột embeddings).

df['similarity'] = df['embeddings'].apply(lambda x: cosine_similarity([query], [x])[0][0])
Ý nghĩa: Duyệt qua từng hàng trong cột df['embeddings'] để tính độ tương đồng giữa vector truy vấn (query) và vector của từng sản phẩm (x).

cosine_similarity([query], [x]): Hàm của scikit-learn nhận vào các mảng 2D và trả về một ma trận độ tương đồng.

[0][0]: Lấy ra giá trị số thực đơn lẻ từ ma trận kết quả (giá trị này thường nằm trong khoảng từ -1 đến 1, giá trị càng gần 1 nghĩa là sản phẩm càng khớp nghĩa với từ khóa tìm kiếm).

# Gradio visualize


```python
import gradio as gr
import pandas as pd
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# 1. Tải lại model và dữ liệu đã lưu trước đó
print("Đang tải mô hình và dữ liệu...")
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
df = pd.read_pickle('product_embeddings.pkl')
print("Đã sẵn sàng khởi chạy ứng dụng!")

# 2. Hàm tìm kiếm trả về định dạng phù hợp cho UI đẹp mắt
def search_products(query, top_k):
    if not query.strip():
        return [["Vui lòng nhập từ khóa tìm kiếm!", "", "", ""]]

    # Mã hóa câu truy vấn
    query_vector = model.encode(query)

    # Tính độ tương đồng
    similarities = cosine_similarity([query_vector], list(df['embeddings']))[0]

    # Tạo bản sao tạm để không ảnh hưởng DataFrame gốc
    temp_df = df.copy()
    temp_df['similarity'] = similarities

    # Sắp xếp và lấy top kết quả
    results = temp_df.sort_values(by='similarity', ascending=False).head(int(top_k))

    # Chuẩn bị dữ liệu hiển thị dạng bảng (format điểm similarity thành phần trăm cho dễ đọc)
    output_data = []
    for _, row in results.iterrows():
        match_percent = f"{row['similarity'] * 100:.1f}%"
        output_data.append([
            row.get('title', 'Không có tiêu đề'),
            row.get('brand', 'Không rõ'),
            row.get('category', 'Chưa phân loại'),
            match_percent
        ])

    return output_data

# 3. Thiết kế giao diện Gradio với Blocks (Modern UI/UX)
with gr.Blocks(theme=gr.themes.Soft(primary_hue="blue", secondary_hue="indigo")) as demo:

    # Tiêu đề và mô tả ứng dụng
    gr.Markdown(
        """
        # 🛍️ Smart E-Commerce Semantic Search
        *Hệ thống tìm kiếm và gợi ý sản phẩm thông minh dựa trên AI & Vector Embeddings.*
        """
    )

    with gr.Row():
        with gr.Column(scale=4):
            # Ô nhập từ khóa tìm kiếm
            query_input = gr.Textbox(
                label="Bạn đang tìm kiếm sản phẩm gì?",
                placeholder="Ví dụ: wireless bluetooth headphones, running shoes, smartphone...",
                lines=1
            )
        with gr.Column(scale=1):
            # Thanh trượt chọn số lượng kết quả
            top_k_slider = gr.Slider(
                minimum=1, maximum=20, value=5, step=1,
                label="Số lượng kết quả"
            )

    # Nút bấm tìm kiếm chính
    search_btn = gr.Button("🔍 Tìm kiếm sản phẩm", variant="primary")

    # Gợi ý nhanh từ khóa (Click vào để test nhanh)
    gr.Examples(
        examples=[
            ["wireless bluetooth headphones", 5],
            ["running shoes for men", 5],
            ["smart watch fitness tracker", 5],
            ["organic face cream", 5]
        ],
        inputs=[query_input, top_k_slider]
    )

    # Khu vực hiển thị kết quả dạng Bảng (Dataframe) trực quan
    gr.Markdown("### 📊 Kết quả phù hợp nhất")
    output_table = gr.Dataframe(
        headers=["Tên sản phẩm", "Thương hiệu", "Danh mục", "Độ khớp (Similarity)"],
        datatype=["str", "str", "str", "str"],
        row_count=5,
        col_count=(4, "fixed"),
        interactive=False
    )

    # Xử lý sự kiện khi bấm nút hoặc nhấn Enter ở ô input
    search_btn.click(
        fn=search_products,
        inputs=[query_input, top_k_slider],
        outputs=output_table
    )

    query_input.submit(
        fn=search_products,
        inputs=[query_input, top_k_slider],
        outputs=output_table
    )

# 4. Chạy ứng dụng
if __name__ == "__main__":
    demo.launch(server_name="127.0.0.1", server_port=7860, share=False)
```
