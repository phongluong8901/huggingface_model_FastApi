```python
# pip install -U transformers accelerate
# Load model directly
from transformers import AutoImageProcessor, AutoModelForImageClassification

# Sử dụng processor của mô hình tương thích vì repository gốc thiếu preprocessor_config.json
processor = AutoImageProcessor.from_pretrained("google/vit-base-patch16-224")
model = AutoModelForImageClassification.from_pretrained("Folefac/vit-plant", device_map="auto")
```

```python
import torch

# 1. Tạo dữ liệu mẫu (dummy input) chuẩn kích thước ViT (1 ảnh, 3 kênh màu, 224x224 pixels)
dummy_input = torch.randn(1, 3, 224, 224)

# 2. Định nghĩa tên đầu vào và đầu ra
input_names = ["pixel_values"]
output_names = ["logits"]

# 3. Tiến hành xuất sang file ONNX
torch.onnx.export(
    model,
    dummy_input,
    "vit_classification.onnx",
    export_params=True,
    opset_version=14,
    do_constant_folding=True,
    input_names=input_names,
    output_names=output_names,
    dynamic_axes={
        "pixel_values": {0: "batch_size"},
        "logits": {0: "batch_size"}
    }
)

print("Đã export sang file 'vit_classification.onnx' thành công!")
```

```python
from google.colab import files
import onnxruntime as ort
import numpy as np
from PIL import Image
import io

# 1. Mở ô chọn file từ máy tính của bạn
uploaded = files.upload()

# 2. Lấy tên file vừa tải lên
for fn in uploaded.keys():
    print(f"Đã tải lên thành công file: {fn}")
    img_path = fn

# 3. Khởi tạo phiên làm việc ONNX Runtime
session = ort.InferenceSession("vit_classification.onnx", providers=["CPUExecutionProvider"])

# 4. Đọc và tiền xử lý ảnh bạn vừa chọn
test_img = Image.open(img_path).convert("RGB")
resized_img = test_img.resize((224, 224))

np_array = np.asarray(resized_img, dtype=np.float32) / 255.0
img = np.transpose(np_array, (2, 0, 1))
img = np.expand_dims(img, axis=0)

# 5. Chạy suy luận ONNX
input_name = session.get_inputs()[0].name
session_input = {input_name: img}

onnx_output = session.run(None, session_input)
onnx_logits = onnx_output[0]

# 6. In kết quả chỉ số lớp dự đoán
onnx_pred_idx = np.argmax(onnx_logits, axis=1)[0]
print(f"Chỉ số lớp dự đoán cho ảnh của bạn: {onnx_pred_idx}")
```

```python
# 1. Cài đặt thư viện Gradio (nếu chưa có)
!pip install -q gradio

import gradio as gr
import onnxruntime as ort
import numpy as np
from PIL import Image

# 2. Khởi tạo phiên làm việc ONNX (nếu session chưa chạy sẵn)
session = ort.InferenceSession("vit_classification.onnx", providers=["CPUExecutionProvider"])

# Lấy từ điển nhãn từ mô hình gốc (id2label)
# Nếu không có sẵn, bạn có thể tự định nghĩa dictionary theo đúng thứ tự các lớp của bạn
labels = model.config.id2label if hasattr(model, 'config') and hasattr(model.config, 'id2label') else {0: 'Healthy', 1: 'Powdery', 2: 'Rust'}

# 3. Hàm xử lý ảnh và dự đoán cho Gradio
def predict_plant_disease(input_image):
    if input_image is None:
        return "Vui lòng tải lên một bức ảnh!"

    # Resize và chuẩn hóa ảnh
    resized_img = input_image.convert("RGB").resize((224, 224))
    np_array = np.asarray(resized_img, dtype=np.float32) / 255.0
    img = np.transpose(np_array, (2, 0, 1))
    img = np.expand_dims(img, axis=0)

    # Chạy suy luận ONNX
    input_name = session.get_inputs()[0].name
    onnx_output = session.run(None, {input_name: img})
    logits = onnx_output[0][0]

    # Tính toán xác suất (Softmax) để hiển thị tỉ lệ % độ tin cậy
    exp_logits = np.exp(logits - np.max(logits))
    probabilities = exp_logits / np.sum(exp_logits)

    # Đóng gói kết quả dạng từ điển {tên_bệnh: tỉ_lệ} cho Gradio Label
    confidences = {labels[i]: float(probabilities[i]) for i in range(len(labels))}
    return confidences

# 4. Xây dựng giao diện Web bằng Gradio
demo = gr.Interface(
    fn=predict_plant_disease,
    inputs=gr.Image(type="pil", label="Tải lên hoặc kéo thả ảnh lá cây cần kiểm tra"),
    outputs=gr.Label(num_top_classes=3, label="Kết quả dự đoán bệnh & Độ tin cậy"),
    title="🌿 Hệ thống phân loại bệnh lá cây (ViT + ONNX Runtime)",
    description="Mô hình Vision Transformer chạy siêu tốc qua ONNX Runtime trên Google Colab.",
    flagging_mode="never"
)

# 5. Khởi chạy ứng dụng (share=True để tạo link công khai truy cập được từ bên ngoài)
demo.launch(share=True, debug=True)
```
